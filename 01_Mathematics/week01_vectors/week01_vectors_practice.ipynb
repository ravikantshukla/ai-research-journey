# Week 1 — Vectors, Linear Combinations, and Span

## Recap

- **Vector**: an ordered list of numbers, e.g. `[3, 2]`. Geometrically, it's an arrow from the origin `(0, 0)` to the point `(3, 2)`. It has both a *direction* and a *length* (magnitude).
- **Linear combination**: given vectors `v` and `w` and scalars `a` and `b`, the expression `a*v + b*w` is a linear combination. It means "scale each vector, then add them tip-to-tail."
- **Span**: the set of *all* possible linear combinations of a set of vectors. The span of one nonzero 2D vector is a line through the origin. The span of two non-parallel 2D vectors is the entire 2D plane. If two vectors *are* parallel (one is a scalar multiple of the other), their span collapses back down to just a line.

Keep these three ideas connected: a linear combination is *one* point you can reach by scaling and adding; the span is *every* point you could ever reach that way.

## Demos

Run each cell and read the comment — it explains what's happening geometrically, not just the code.

```python
import numpy as np
import numpy as np
u = np.array([2, 1])
u = np.array()
v = np.array([1, 3])

u_plus_v = u + v
print("u + v =", u_plus_v)  # tip-to-tail: start at origin, walk u, then walk v from there
```

```python
c = 2.5
scaled_u = c * u
print(f"{c} * u =", scaled_u)  # stretches u to 2.5x its length, same direction (flips direction if c is negative)
```

```python
dot = u @ v  # equivalent to np.dot(u, v)
print("u . v =", dot)  # measures how much u and v point in the same direction (0 = perpendicular)
```

```python
norm_u = np.linalg.norm(u)
print("||u|| =", norm_u)  # the length of vector u (distance from origin to the tip of u)
```

```python
a, b = 3, -1
combo = a * u + b * v
print(f"{a}*u + {b}*v =", combo)  # one specific point reachable by scaling u and v, then adding them
```
