## 1. Variables, Types & Operators

Python is **dynamically typed**: names are labels bound to objects, and the *object* has the type.
Core scalar types: `int` (arbitrary precision), `float` (64-bit), `complex`, `bool`, `str`, `NoneType`.

```python
a = 42            # int
b = 3.14          # float
c = 2 + 3j        # complex
d = True          # bool (subclass of int!)
e = None          # the "no value" singleton
f = "hello"       # str

for value in (a, b, c, d, e, f):
    print(f"{value!r:>10} -> {type(value).__name__}")

print("bool is int subclass:", isinstance(True, int), "| True + True =", True + True)
```

### Arithmetic operators
| Op | Meaning | Example |
|---|---|---|
| `/` | true division (always float) | `7 / 2 == 3.5` |
| `//` | floor division | `7 // 2 == 3`, `-7 // 2 == -4` |
| `%` | modulo (sign follows divisor) | `-7 % 3 == 2` |
| `**` | power | `2 ** 10 == 1024` |
| `divmod(a, b)` | `(a // b, a % b)` | `divmod(17, 5) == (3, 2)` |

```python
print(7 / 2, 7 // 2, -7 // 2, 7 % 3, -7 % 3, 2 ** 10, divmod(17, 5))
print("Big ints are exact:", 2 ** 100)
print("Float precision gotcha:", 0.1 + 0.2, "==", 0.1 + 0.2 == 0.3)

import math
print("Use math.isclose:", math.isclose(0.1 + 0.2, 0.3))
print("round() uses banker's rounding:", round(2.5), round(3.5), round(3.14159, 2))
```

### Comparison, logical & identity operators
- `==` compares **values**; `is` compares **identity** (same object). Use `is` only for `None`, `True`, `False`.
- `and` / `or` return one of their operands (short-circuit), not necessarily a `bool`.
- Chained comparisons: `0 < x < 10`.

```python
x = 5
print(0 < x < 10)                 # chained
print([] or "default")            # 'or' returns first truthy operand
print("first" and "second")       # 'and' returns last operand if all truthy
print(None is None, [1] == [1], [1] is [1])

# Truthiness: these are all False
falsy = [0, 0.0, "", [], {}, set(), (), None, False]
print([bool(v) for v in falsy])
```

## 2. Strings
Strings are **immutable** sequences of Unicode characters. Index with `s[i]`, slice with `s[start:stop:step]`.

```python
s = "Machine Learning"
print(s[0], s[-1], s[0:7], s[8:], s[::-1], s[::2])
print(len(s), s.lower(), s.upper(), s.title(), s.swapcase())
print(s.startswith("Mach"), s.endswith("ing"), "Learn" in s)
print(s.find("Learn"), s.find("zzz"), s.count("n"))
print(s.replace("Machine", "Deep"))
print(s.split(), "a,b,,c".split(","), " | ".join(["x", "y", "z"]))
print("  padded  ".strip() + "|", "xxhixx".strip("x"), "42".zfill(5))
print("abc123".isalnum(), "123".isdigit(), "abc".isalpha(), "   ".isspace())
```

### f-strings — the formatting mini-language
`f"{value:[fill][align][width][,][.precision][type]}"`

| Spec | Result |
|---|---|
| `{x:.3f}` | 3 decimals |
| `{x:10.2f}` | width 10, right aligned |
| `{x:<10}` / `{x:^10}` / `{x:>10}` | left / center / right |
| `{x:,}` | thousands separator |
| `{x:.2%}` | percentage |
| `{x:e}` | scientific |
| `{x!r}` | use `repr()` |
| `{x=}` | debug: prints `x=value` (3.8+) |

```python
pi = 3.14159265
acc = 0.93456
loss = 0.000123
big = 1234567.891
name = "resnet"
print(f"{pi:.3f} | {pi:10.2f}| {name:<8}| {name:^10}| {name:>8}|")
print(f"{big:,.2f} | {acc:.2%} | {loss:.2e} | {name!r} | {pi=:.2f}")
print(f"{'epoch':<6}{'loss':>10}{'acc':>8}")
for ep, (l, a_) in enumerate([(0.9, 0.61), (0.5, 0.80), (0.21, 0.93)], start=1):
    print(f"{ep:<6}{l:>10.4f}{a_:>8.1%}")
```

### Loops: `for`, `while`, `break`, `continue`, `else`
The `else` on a loop runs only if the loop **didn't** `break`.

```python
for i in range(0, 10, 3):
    print(i, end=" ")
print()

for n in range(2, 20):
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            break
    else:                         # no divisor found
        print(n, end=" ")         # primes
print()

count = 0
while True:
    count += 1
    if count % 2 == 0:
        continue
    if count > 7:
        break
    print("odd count", count)
```

### Structural pattern matching (`match`, Python 3.10+)

```python
def describe(obj):
    match obj:
        case 0:
            return "zero"
        case int() | float() if obj < 0:
            return "negative number"
        case [x, y]:
            return f"pair of {x} and {y}"
        case {"type": "conv", "filters": f}:
            return f"conv layer with {f} filters"
        case str() as s:
            return f"string of length {len(s)}"
        case _:
            return "something else"

for item in (0, -3.2, [1, 2], {"type": "conv", "filters": 64}, "hey", {1, 2}):
    print(f"{item!r:35} -> {describe(item)}")
```

## 5. Comprehensions
Syntax: `[expr for item in iterable if cond]`. Also dict `{k: v ...}`, set `{x ...}`, and generator `(x ...)`.

```python
squares = [x ** 2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]
labels = ["pos" if x > 0 else "neg" for x in (-2, 3, -1)]
flat = [v for row in [[1, 2], [3, 4], [5]] for v in row]      # nested loops: outer first
matrix_T = [[row[i] for row in [[1, 2, 3], [4, 5, 6]]] for i in range(3)]
word_len = {w: len(w) for w in ["numpy", "pandas", "torch"]}
inverted = {v: k for k, v in word_len.items()}
unique_lens = {len(w) for w in ["a", "bb", "cc", "ddd"]}
gen = (x * x for x in range(1_000_000))   # lazy, no big list built
print(squares, evens, labels, flat, matrix_T, word_len, inverted, unique_lens, sep="\n")
print("sum of 1M squares via generator:", sum(gen))
```

## 6. Functions

Parameter order: `def f(pos_only, /, normal, *args, kw_only, **kwargs)`
- Default values are evaluated **once** at definition time → never use a mutable default like `[]`.

```python
def train(model, epochs=10, *callbacks, lr=0.01, verbose=False, **extra):
    """Demonstrates every kind of parameter."""
    print(f"model={model}, epochs={epochs}, callbacks={callbacks}, lr={lr}, verbose={verbose}, extra={extra}")

train("mlp")
train("cnn", 5, "early_stop", "checkpoint", lr=0.1, momentum=0.9)

def f(a, b, /, c, *, d):          # a,b positional-only; d keyword-only
    return a + b + c + d
print(f(1, 2, c=3, d=4))
```

```python
# Mutable default gotcha
def bad_append(x, acc=[]):
    acc.append(x); return acc
def good_append(x, acc=None):
    acc = [] if acc is None else acc
    acc.append(x); return acc
print(bad_append(1), bad_append(2))    # shared list!
print(good_append(1), good_append(2))
```

```python
# Unpacking into calls
def area(w, h): return w * h
dims = (3, 4); kw = {"w": 5, "h": 6}
print(area(*dims), area(**kw))

# Returning multiple values = returning a tuple
def stats(xs):
    return min(xs), max(xs), sum(xs) / len(xs)
lo, hi, mean = stats([4, 8, 15, 16, 23, 42])
print(lo, hi, round(mean, 2))

# Lambdas: small anonymous functions
square = lambda z: z * z
pairs = [(1, "b"), (3, "a"), (2, "c")]
print(square(5), sorted(pairs, key=lambda p: p[1]))

# Type hints + docstring
def normalize(values: list[float], eps: float = 1e-8) -> list[float]:
    """Scale values to [0, 1]."""
    lo, hi = min(values), max(values)
    return [(v - lo) / (hi - lo + eps) for v in values]
print(normalize([2.0, 4.0, 6.0]))
print(normalize.__doc__, normalize.__annotations__)
```

### Scope: LEGB (Local → Enclosing → Global → Built-in), `global`, `nonlocal`, closures

```python
counter = 0
def inc_global():
    global counter
    counter += 1

def make_counter():
    count = 0
    def inc():
        nonlocal count
        count += 1
        return count
    return inc                       # closure remembers `count`

inc_global(); inc_global()
c1 = make_counter()
print(counter, c1(), c1(), c1())
```

## 8. Exceptions
`try / except / else / finally`, raising, custom exceptions, and `assert`.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError as err:
        print("caught:", err)
        return float("nan")
    except TypeError:
        print("bad types!")
        raise                               # re-raise
    else:
        print("no error")                   # runs only if no exception
        return result
    finally:
        print("finally always runs")

print(safe_divide(6, 3)); print(safe_divide(1, 0))

class ShapeMismatchError(ValueError):
    """Custom error for shape problems."""

def check_shapes(a, b):
    if len(a) != len(b):
        raise ShapeMismatchError(f"{len(a)} != {len(b)}")
try:
    check_shapes([1, 2], [1])
except ShapeMismatchError as e:
    print(type(e).__name__, "->", e)

try:
    assert 1 + 1 == 3, "math is broken"
except AssertionError as e:
    print("AssertionError:", e)
```

## 10. Standard Library Essentials for ML Work

```python
from collections import Counter, defaultdict, deque, OrderedDict

words = "the cat and the hat and the bat".split()
cnt = Counter(words)
print(cnt.most_common(2), cnt["the"], cnt["zzz"])

groups = defaultdict(list)
for w in words:
    groups[len(w)].append(w)
print(dict(groups))

window = deque(maxlen=3)          # rolling window (e.g. last 3 losses)
for loss in [0.9, 0.7, 0.6, 0.4, 0.35]:
    window.append(loss)
print(window, sum(window) / len(window))
```

```python
import itertools as it
print(list(it.product([1, 2], ["a", "b"])))          # grid search combos
print(list(it.combinations("ABC", 2)), list(it.permutations("AB")))
print(list(it.chain([1, 2], [3], [4, 5])))
print(list(it.islice(it.count(10, 5), 4)))
print([(k, list(g)) for k, g in it.groupby("aaabbc")])
print(list(it.accumulate([1, 2, 3, 4])))
# Hyper-parameter grid
grid = {"lr": [0.1, 0.01], "bs": [16, 32]}
for combo in it.product(*grid.values()):
    print(dict(zip(grid.keys(), combo)))
```

```python
from functools import reduce, partial, lru_cache
print(reduce(lambda acc, v: acc * v, [1, 2, 3, 4, 5]))
pow2 = partial(pow, 2)
print(pow2(10))

@lru_cache(maxsize=None)
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)
print(fib(80), fib.cache_info())
```

```python
s = "Deep Learning with PyTorch"
print(s[::-1], sum(ch in "aeiou" for ch in s.lower()))

text = "a quick brown fox jumps over a lazy dog".split()
freq = {}
for w in text:
    freq[w] = freq.get(w, 0) + 1
print(freq == dict(Counter(text)), freq)

def chunk(lst, n):
    return [lst[i:i + n] for i in range(0, len(lst), n)]
print(chunk(list(range(7)), 3))

def flatten(nested):
    out = []
    for item in nested:
        out.extend(flatten(item) if isinstance(item, list) else [item])
    return out
print(flatten([1, [2, [3, [4]], 5]]))

scores = {"alice": 88, "bob": 95, "cara": 72}
for rank, (name, sc) in enumerate(sorted(scores.items(), key=lambda kv: -kv[1]), 1):
    print(f"{rank}. {name} ({sc})")

def safe_int(s, default=0):
    try:
        return int(s)
    except (ValueError, TypeError):
        return default
print(safe_int("12"), safe_int("x"), safe_int(None, -1))

print(sum(1 for a, b in it.product(range(1, 7), repeat=2) if a + b == 7))
```
